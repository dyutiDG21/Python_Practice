## Exercise 25: Mini Project Challenge (Notebook Version)

Build a simple number guessing checker. Because notebooks handle `input()` awkwardly, use a fixed `secret` and a list of `guesses`.

1. Set `secret` to a number between 1 and 20.
2. Create a list called `guesses` with a few numbers.
3. Loop through `guesses` and for each one print whether it is too high, too low, or correct.
4. Stop the loop once a guess is correct and print how many guesses it took.

```python
secret = 13
guesses = [10, 18, 13]
# Your code here
guess_count = 0
for guess in guesses:
    guess_count += 1
    if guess < secret:
        print(guess, "is too low.")
    elif guess > secret:
        print(guess, "is too high.")
    else:
        print(guess, "is correct")
        print(f"It took {guess_count} guesses.")
        break
```

## Exercise 26: Mini Project Challenge

Build a simple to-do list using a list of commands instead of live input.

1. Create an empty list called `tasks`.
2. Create a list called `commands`, for example `["add Buy milk", "add Walk dog", "view"]`.
3. Loop through `commands`. If a command starts with `add `, store the rest of the text in `tasks`. If it is `view`, print each task with a number next to it.
4. At the end, print how many tasks were added.

```python
tasks = []
commands = ["add Buy milk", "add Walk dog", "view"]
# Your code here
for command in commands:
    if command.startswith("add "):
        task = command[4:]
        tasks.append(task)
    elif command == "view":
        print("Tasks:")
        for number, task in enumerate(tasks, start=1):
            print(number, task)
print("Number of tasks added:", len(tasks))
```
